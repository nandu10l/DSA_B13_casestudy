```python
# import pandas, numpy
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("Numpy version:", np.__version__)
print("____________________________________________________________________________________________________________\n")
# loading dataset usind pd.read_csv()
filepath = 'https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv'
df = pd.read_csv(filepath)

# to display shape, head, column names, datatypes, missing values
print("SHAPE \n")
print(df.shape)
print("____________________________________________________________________________________________________________\n")
print("\nHEAD \n")
print(df.head())
print("____________________________________________________________________________________________________________\n")
print("\nCOLUMN NAMES\n")
print(df.columns)
print("____________________________________________________________________________________________________________\n")
print("\nDATATYPES \n")
print(df.dtypes)
print("____________________________________________________________________________________________________________\n")
print("\n MISSING VALUE \n")
print(df.isnull().sum())

print("____________________________________________________________________________________________________________\n")
# to check and remove duplicates
print("\nDUPLICATES \n")
if df.duplicated().sum() > 0:
  print("Number of duplicate rows: ",df.duplicated().sum())
  df.drop_duplicates(inplace=True)
else:
  print("No duplicates\n")
print("____________________________________________________________________________________________________________\n")
# to rename column ("Title" > "Movie_Title")
df = df.rename(columns={"Title":"Movie_Title"})
print(df.columns,"\n")
print("____________________________________________________________________________________________________________\n")
# to get unique genres
print("UNIQUE GENRES\n")
genres = sorted({genre.strip() for cell in df['Genre'].dropna() for genre in cell.split(',')})
print('Unique genres:', genres)

print("____________________________________________________________________________________________________________\n")
# to extract number of genres per movie
def count_genres(value):
    if pd.isna(value):
        return 0
    return len(value.split(","))

df["Genre_Count"] = df["Genre"].apply(count_genres)
print("\nNUMBER OF GENRES PER MOVIE\n")
print(df.Genre_Count,"\n")
print("____________________________________________________________________________________________________________\n")
# to handle missing value in 'Revenue(Millions)' and 'Metascore'
for column in ['Revenue (Millions)', 'Metascore']:
    missing_before = df[column].isna().sum()
    median_value = df[column].median()
    df[column] = df[column].fillna(median_value)
    print(f'{column}: filled {missing_before} missing values with median {median_value:.2f}')

print("____________________________________________________________________________________________________________\n")
# to print top 5 movies by rating
print("\n")
top_5 = df.sort_values("Rating", ascending=False).head(5)
print(top_5[["Movie_Title", "Year", "Rating", "Revenue (Millions)"]])

print("____________________________________________________________________________________________________________\n")
# to print movies with highest/lowest revenue
highest_revenue = df.sort_values("Revenue (Millions)", ascending=False).head(1)
lowest_revenue = df.sort_values("Revenue (Millions)", ascending=True).head(1)
print("\n")
print("Highest Revenue:")
print(highest_revenue[["Movie_Title", "Year", "Revenue (Millions)"]])
print("\n")
print("Lowest Revenue:")
print(lowest_revenue[["Movie_Title", "Year", "Revenue (Millions)"]])

print("____________________________________________________________________________________________________________\n")
# to print average rating by Director
director_rating = df.groupby("Director")["Rating"].mean()
print("\n")
print("Average Rating by Director:")
print(director_rating)
print("____________________________________________________________________________________________________________\n")
# to print movies per year
movies_per_year = df.groupby("Year")["Movie_Title"].count()
print("\n")
print("Movies per Year:")
print(movies_per_year)
print("____________________________________________________________________________________________________________\n")
# to print movies with rating > 8 and revenue > 100M
filtered_movies = df[(df["Rating"] > 8) & (df["Revenue (Millions)"] > 100)]
print("\n")
print("Movies with Rating > 8 and Revenue > 100M:")
print(filtered_movies)
print("____________________________________________________________________________________________________________\n")
# to group by Director: avg rating, Revenue
director_stats = df.groupby("Director").agg({"Rating": "mean", "Revenue (Millions)": "sum"})
print("\n")
print("Director Statistics:")
print(director_stats)
print("____________________________________________________________________________________________________________\n")
# group by Year: count,avg_rating
year_stats = df.groupby("Year").agg({"Movie_Title": "count", "Rating": "mean"})
print("\n")
print("Year Statistics:")
print(year_stats)

print("____________________________________________________________________________________________________________\n")
# to create new column "Profitability" = revenue divided by duration

df["Profitability"] = df["Revenue (Millions)"] / df["Runtime (Minutes)"]
print("\n")
print("Profitability Column:")
print(df["Profitability"])
print("____________________________________________________________________________________________________________\n")
# to print the top 10 most profitable movies
most_profitable = df.sort_values("Profitability", ascending=False).head(10)
print("\n")
print("Top 10 Most Profitable Movies:")
print(most_profitable[["Movie_Title", "Year", "Profitability"]])

print("____________________________________________________________________________________________________________\n")
# to create a columns that categorizes movies by rating :
# rating < 6:Poor
# 6 <= Rating < 7.5 : Average
# Rating >= 7.5 : Excellent

def categorize_rating(rating):
    if rating < 6:
        return "Poor"
    elif 6 <= rating < 7.5:
        return "Average"
    else:
        return "Excellent"

df["Rating_Category"] = df["Rating"].apply(categorize_rating)
print("\n")
print("Rating Category Column:")
print(df["Rating_Category"])

print("____________________________________________________________________________________________________________\n")
# custom function to score movies based on:
  # Apply row['Rating'] * 10 + row['Revenue (Millions)'] as High score

def custom_score(row):
    return row['Rating'] * 10 + row['Revenue (Millions)']

df['Custom_Score'] = df.apply(custom_score, axis=1)
print("\n")
print("Custom Score Column:")
print(df["Custom_Score"])
```

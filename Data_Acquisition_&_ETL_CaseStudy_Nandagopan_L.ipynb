# PART_C

```python
#TASK-6
# Rename columns wherever necessary so that all datasets have a common column name for merging.

df_xlsx = df_xlsx.rename(columns={'Country': 'country'})
df_xlsx.head()

df_xml = df_xml.rename(columns={'name': 'country'})
df_xml.head()
```

```python
#TASK-7
# Convert the required columns into appropriate numeric data types

```

```python
#TASK-8
# Merge all four datasets into one DataFrame.

df_merged = pd.merge(df, df_xlsx, on='country')
df_merged = pd.merge(df_merged, df_json, on='country')
df_merged = pd.merge(df_merged, df, on='country')
df_merged = pd.merge(df_merged,df_xml, on='country')
df_merged.head()
```

```python
#TASK-9
# Create the following calculated columns: 1. Internet Penetration (%), 2. GDP Per Capita

df_merged['Internet_Penetration'] = (df_merged['internet_users'] / df_merged['population_x']) * 100
df_merged['GDP_Per_Capita'] = df_merged['GDP'] / df_merged['population_x']
df_merged.head()
```

```python
#TASK-10
# Create a SQLite database. Store the final merged dataset into a table named ‘country_statistics’ Hint: ● Use the sqlite3 library. ● Load the DataFrame using pandas.

import sqlite3

conn = sqlite3.connect('country_statistics.db')
df_merged.to_sql('country_statistics', conn, if_exists='replace', index=False)
pd.read_sql_query('SELECT * FROM country_statistics', conn)
```

```python
#TASK-11
# to verify that the data has been successfully loaded
pd.read_sql_query('SELECT * FROM country_statistics', conn).head()
```

## Summary of the ETL and Analysis Process
### Diverse Data Sources:
We extracted data from four distinct file formats: a CSV file for population records (population.csv), an Excel spreadsheet for GDP statistics (gdp.xlsx), a JSON file for internet user counts (internet_users.json), and an XML document for literacy rates (literacy_rate.xml).
### Targeted Data Cleaning:
Key column names were standardized (e.g., renaming 'Country' to 'country' and 'name' to 'country') to establish a uniform key for merging, and we validated that there were no duplicate entries or missing values across the tables.
### Comprehensive Transformations:
All four datasets were merged on the 'country' column into a unified DataFrame. We then calculated two custom metrics: Internet Penetration (%) (internet users divided by population) and GDP Per Capita (GDP divided by population).
### Robust Database Loading:
We established a connection to a local SQLite database and created a table named country_statistics using Python's sqlite3 and pandas libraries, successfully loading all merged data for SQL-based querying.
Key Analytical Insights:
### GDP & Population:
The United States has the highest total GDP, while China has the largest population in our dataset.
### Development Indicators:
Switzerland boasts the highest GDP per capita, and Norway has the highest internet penetration (97.76%) coupled with a 100% literacy rate.
### Correlation:
We found a strong positive correlation (~0.77) between a country's literacy rate and its internet penetration percentage.

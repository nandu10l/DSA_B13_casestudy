```python
employees = [
    {"id":101, "name":"Alice", "department":"Data", "salary":65000, "rating":4.5},
    {"id":102, "name":"Bob", "department":"AI", "salary":72000, "rating":4.8},
    {"id":103, "name":"Charlie", "department":"Web", "salary":58000, "rating":3.9},
    {"id":104, "name":"David", "department":"Data", "salary":61000, "rating":4.2},
    {"id":105, "name":"Eva", "department":"AI", "salary":75000, "rating":4.9},
    {"id":106, "name":"Frank", "department":"Cloud", "salary":68000, "rating":4.1},
    {"id":107, "name":"Grace", "department":"Web", "salary":59000, "rating":3.8},
    {"id":108, "name":"Helen", "department":"Data", "salary":70000, "rating":4.6}
]

    # input taking employee id
employee_id=input("Enter Employee ID:")

print("\n")
#Employee details of employee id entered
for emp in employees:
  if emp["id"]==int(employee_id):


#displaying the total no. of employees
    print("Total Employees :",len(employees))
    print("\n")

#Display the average, highest, and lowest salary.
    total_salary=0
    for emp in employees:
      total_salary+=emp["salary"]

#average salary
    print("Average Salary:",total_salary/len(employees))
#highest salary
    print("Highest Salary:",max(emp["salary"] for emp in employees))
#lowest salary
    print("Lowest Salary:",min(emp["salary"] for emp in employees))

    print("\n")
#Display the names of employees whose salary is greater than ₹65,000 and whose performance rating is 4.5 or above.
    print("Employees with Salary>65000 and Rating>=4.5")
    for emp in employees:
      if emp["salary"]>65000 and emp["rating"]>=4.5:
        print(emp["name"])
      else:
        continue

#dictionary with Employee ID as the key and Employee Name as the value
    print("\nEmployee Dictionary")
    for emp in employees:
      emp_dict={emp["id"]:emp["name"]}
      print(emp_dict)

    print("\n")
#list with all employee names
    print("First Three Employee Names\n")
    emp_names=[emp["name"] for emp in employees]
    emp_names.sort()
    emp_names.reverse()
    print(emp_names[0])
    print(emp_names[1])
    print(emp_names[2])


    print("\n")
#Employee ID Tuple
    print("Employee ID Tuple")
    emp_id_tuple=tuple(emp["id"] for emp in employees)
    print(emp_id_tuple)
    print("\n")
    print("Maximum Employee ID: ",max(emp_id_tuple))
    print("Minimum Employee ID: ",min(emp_id_tuple))

    print("\n")
#Set of departments
    print("Departments")
    departments=set(emp["department"] for emp in employees)
    print(departments)


    print("\n")



    print("Employee Found")
    print("ID:",emp["id"])
    print("Name:",emp["name"])
    print("Department:",emp["department"])
    print("Salary:",emp["salary"])
    print("Rating:",emp["rating"])




    break
else:
  print("Employee not found")
```
